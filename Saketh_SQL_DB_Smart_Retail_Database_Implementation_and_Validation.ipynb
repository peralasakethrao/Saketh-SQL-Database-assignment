# **Data insertion and population**
The synthetic retail data was put into the database before it was turned into an SQLite database file. Then, I put the old database into Google Colab, where I used Python's SQLite3 library to connect to it. I ran the following code to make sure that the data was put into the database correctly by checking how many records are in every table.

```python
tables = [
    "Customers", "Suppliers", "Categories",
    "Stores", "Products", "Inventory",
    "Orders", "Order_Items", "Payments"
]

total_records = 0

for table in tables:
    count = connection.execute(
        f'SELECT COUNT(*) FROM "{table}"'
    ).fetchone()[0]

    total_records += count
    print(f"{table}: {count} records")

print("\nTotal records:", total_records)
```

# **Primary and foreign key relationships**
The composite primary keys are defined in the Inventory and Order_Items tables. The primary keys provide a unique identifier for each record in every table, whereas the foreign keys create relationships between the related tables. The following SQL statement will retrieve the primary key definitions and foreign key relationships from the SQLite database.

```python
for table in tables:
    print(f"\nTABLE: {table}")

    columns = connection.execute(
        f'PRAGMA table_info("{table}")'
    ).fetchall()

    for column in columns:
        if column[5] > 0:
            print("Primary Key:", column[1])

    foreign_keys = connection.execute(
        f'PRAGMA foreign_key_list("{table}")'
    ).fetchall()

    for fk in foreign_keys:
        print(f"Foreign Key: {fk[3]} -> {fk[2]}.{fk[4]}")
```

# **Data validation and record counts**
Database Validation performs record counts and foreign key checks. These help find records that are missing, as well as foreign keys that are not valid. The following code will determine if there are any foreign key violations present in your database.

```python
connection.execute("PRAGMA foreign_keys = ON")

print("Foreign Key Enforcement:",
      connection.execute("PRAGMA foreign_keys").fetchone()[0])

violations = connection.execute(
    "PRAGMA foreign_key_check"
).fetchall()

if len(violations) == 0:
    print("Foreign key check: No violations found.")
else:
    print("Foreign key violations:", violations)
```
